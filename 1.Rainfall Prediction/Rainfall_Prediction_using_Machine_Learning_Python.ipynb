```python
# Here we can observe that there are unnecessary spaces in the names of the columns let's remove that.

df.rename(str.strip,
          axis='columns',
          inplace=True)

df.columns
```

```python
for col in df.columns:

  # Checking if the column contains
  # any null values
  if df[col].isnull().sum() > 0:
    val = df[col].mean()
    df[col] = df[col].fillna(val)

df.isnull().sum().sum()

# Now it's time for null value imputation.
```

```python
df.replace({'yes':1, 'no':0}, inplace=True)
# There are outliers in the data but sadly we do not have much data so, we cannot remove this
```

```python
'''Sometimes there are highly correlated features that just increase the dimensionality of the feature 
space and do not good for the model's performance. So we must check whether there are highly correlated 
features in this dataset or not.'''


plt.figure(figsize=(10,10))
sb.heatmap(df.corr() > 0.8,
           annot=True,
           cbar=False)
plt.show()
```

```python
'''Now we will remove the highly correlated features 'maxtemp' and 'mintemp'. But why not temp or dewpoint? 
This is because temp and dewpoint provide distinct information regarding the weather and atmospheric conditions.
'''
df.drop(['maxtemp', 'mintemp'], axis=1, inplace=True)
```

```python
'''Model Training
Now we will separate the features and target variables and split them into training and testing data by using 
which we will select the model which is performing best on the validation data.

'''
features = df.drop(['day', 'rainfall'], axis=1)
target = df.rainfall
```

```python
'''As we found earlier that the dataset we were using was imbalanced so, we will have to balance the 
training data before feeding it to the model.
'''


X_train, X_val, Y_train, Y_val = train_test_split(features,
                                      target,
                                      test_size=0.2,
                                      stratify=target,
                                      random_state=2)

# As the data was highly imbalanced we will
# balance it by adding repetitive rows of minority class.
ros = RandomOverSampler(sampling_strategy='minority',
                        random_state=22)
X, Y = ros.fit_resample(X_train, Y_train)
```

```python
# Normalizing the features for stable and fast training.
scaler = StandardScaler()
X = scaler.fit_transform(X)
X_val = scaler.transform(X_val)

'''
The features of the dataset were at different scales so, normalizing it before training will help us 
to obtain optimum results faster along with stable training.'''
```

```python
models = [LogisticRegression(), XGBClassifier(), SVC(kernel='rbf', probability=True)]

for i in range(3):
  models[i].fit(X, Y)

  print(f'{models[i]} : ')

  train_preds = models[i].predict_proba(X)
  print('Training Accuracy : ', metrics.roc_auc_score(Y, train_preds[:,1]))

  val_preds = models[i].predict_proba(X_val)
  print('Validation Accuracy : ', metrics.roc_auc_score(Y_val, val_preds[:,1]))
  print()
```

```python
'''Model Evaluation
From the above accuracies, we can say that Logistic Regression and support vector classifier are satisfactory 
as the gap between the training and the validation accuracy is low. Let's plot the confusion matrix as well 
for the validation data using the SVC model.'''


import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn import metrics

# Use ConfusionMatrixDisplay.from_estimator instead of plot_confusion_matrix
ConfusionMatrixDisplay.from_estimator(models[2], X_val, Y_val)
plt.show()
```
