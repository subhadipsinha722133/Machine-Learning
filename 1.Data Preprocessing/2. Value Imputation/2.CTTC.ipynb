```python
from sklearn.impute import SimpleImputer

sim=SimpleImputer(strategy="mean")
s=sim.fit_transform(a)
s
new_a=pd.DataFrame(s,columns=[a.keys()])
new_a
```

```python
x=new_a.iloc[:,:-1]
y=new_a.iloc[:,-1:]
```

```python
from sklearn.model_selection import train_test_split
x_tarin,x_test,y_tarin,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
lor=LogisticRegression()
lor.fit(x_tarin,y_tarin)
lor.score(x_test,y_test)*100
```
