```python
q1=a["retail_price"].quantile(0.25)
q2=a["retail_price"].quantile(0.75)

IQR=q2-q1
```

```python
min_rang=q1-(IQR*1.5)
max_range=q2+(IQR*1.5)
```

```python
print(min_rang,max_range)
```

```python
new_a=a[a["retail_price"]<=max_range]
```
