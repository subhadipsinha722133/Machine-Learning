```python


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
a=sns.load_dataset('iris')
a
```

```python
a.info()
```

```python
from sklearn.preprocessing import LabelEncoder
la=LabelEncoder()
a["species"]=la.fit_transform(a["species"])
a
```

```python
a.describe()
```

```python
sns.boxplot(a)
```

```python
q1=a["sepal_width"].quantile(0.25)
q2=a["sepal_width"].quantile(0.75)
```

```python
IQR=q2-q1
```

```python
min_rang=q1-(1.5*IQR)
max_rang=q2+(1.5*IQR)
```

```python
new_a=a[a["sepal_width"]<=max_rang]
a
```

```python
new_a.describe()
```

```python
sns.boxplot(new_a["sepal_width"])
```

```python
sns.boxplot(new_a)
```
